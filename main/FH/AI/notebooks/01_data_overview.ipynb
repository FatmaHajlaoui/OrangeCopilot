# 01 — Data Overview & MLO Parsing

**Goal:** Initial exploration of the 4 core telecom datasets (SANITY, RXLev, FH RSL, RTWP),
plus building and running the parser for the ~3961 Atoll Link Reporter (MLO) files.

**Inputs:** `FH/data/*.xlsx`, `FH/data/temp/*.xlsx`, `FH/data/MLO/*.xlsx`

**Outputs:**
- Confirmed structure, dtypes, missing values, duplicates for all 4 core datasets
- Identified `Sanity` and `Status` as classification targets
- Identified sentinel-value pattern (0 dBm / ~-99 dBm = "no reading")
- `src/preprocessing/mlo_parser.py` — parses one Atoll Link Reporter file into a flat dict
- `data/processed/atoll_links.csv` — 3795 parsed MLO links
- `data/processed/atoll_links_parse_errors.csv` — 166 excluded files (164 Excel lock files, 2 malformed)
